## String methods

1) index(character/substring): returns the position of character in the string. If character is not found, it will throw an error
2) find(character/substring): returns the position of character in the string. If character is not found, it will return -1
3) count(character/substring):returns the number of occurrences of character in the string
4) lower(): converts the string to lowercase
5) upper(): converts the string to uppercase
6) title(): converts the 1st character of every word in the string to uppercase and remaining all characters to lowercase
7) capitalize(): converts the 1st character of 1st word in the string to uppercase and remaining all characters to lowercase
8) replace(oldcharacter,newcharacter): replaces old data with new data in new copy of string

## Escape sequences:

    Add a new line: \n
    Add a backspace: \b
    Add a tabspace: \t
    Add a \ : any character that comes after \ is considered as string only

```python
sentence = "This is a test\n file. Writing this using\tescape sequence"
print(sentence)
```

```python
# sentence = 'This is'nt the right way to do'
sentence1 = "This is'nt the right way to do"
sentence2 = 'This is\'nt the right way to do'
print(sentence1,"\n",sentence2)
```

```python
word = "tests\b"
print(word)
print(word.isprintable()) # return false when escape characters are present
print(name2.isprintable()) # return true when escape characters are NOT present
```

## Input() and split()

Input() : takes input from the user. will always be in string format only

```python
scores_all = input("Enter all scores separated by space: ")
print(scores_all)
```

```python
scores_all2 = input("Enter all scores separated by space: ").split()
print(scores_all2)
```

```python
scores_all3 = input("Enter all scores separated by underscore ").split('_')
# Inputs: 45_78_77
print(scores_all3)
```

```python
scores_all4 = input("Enter all scores separated by hashtag ").split('#')
# Inputs: 90#78#67
print(scores_all4)
```

## Strip(): removes extra whitespaces at the beginning and end of the string
