**Create a function multiply that can take 2inputs and return their product as an output**

and: &
or: |

```python
def multiply(num1:int|float,num2:int|float)->int|float:
    '''
    This function takes 2 numerical inputs(can be integers or float data) and returns
    their product as output.
    Input: int/float
    Output: product of 2 inputs : int/float
    Examples:
    multiply(90,10)
     Output>> 900
    multiply(1.5,4.0)
    output >> 6.0
    multiply(2.5,4)
    output >> 10
    '''
    z = num1*num2
    return round(z,2) # return round(num1*num2,2)
```

**Create a function that takes a list of salaries as input and updates each employee's salary by 5%.
Hint: Create function without return keyword**

```python
def salary_increment_5(salaries:list)->None:
    """
    THis function takes a list of salaries as input.
    Increments each salary amount with 5%.
    Returns nothing.
    Example: salary_increment_5([5000,2000,3000])
    Ouput: Nothing
    But the salaries will be updated with 5% increase.
    """
    for i in range(len(salaries)):
        # i=0,i=1,i=2,....
        salaries[i] = salaries[i] + (0.05*salaries[i])
```

```python
emp_salaries = [50000,80000,90000,150000]
print(f"Before Increment: {emp_salaries}")
salary_increment_5(emp_salaries)
print(f"After increment: {emp_salaries}")
```

**Write a program that creates a function which takes students scores as input and their name as optional input. Display a message "Congratulations User. Your avg score is ___"**

```python
# emp_salaries.pop()
def student_scores(scores:list,name="User")->str:
    """
    This functions takes students scores as input and their name as optional input.
    Display a message "Congratulations User. Your avg score is ___"
    Examples: student_scores([90,89,78])
    Congratulations User. Your avg score is 85.66
    student_scores([90,89,78],"Raman")
    Congratulations Raman. Your avg score is 85.66
    """
    avg = sum(scores)/len(scores)
    return f"Congratulations {name}. Your avg score is {round(avg,2)}"
```

```python
sc1 = student_scores([90,89,78,67])
sc2 = student_scores([67,54,88,98,78],"Thaman")
print(sc1,sc2,sep='\n\n')
```

```python
def student_scores2(math_score,sci_score,social_score,name="User",section="3G"):
    """
        This functions takes 3 students scores as input and their name as optional input.
        Display a message "Congratulations User. Your avg score is ___"
        Examples: student_scores(90,89,78)
        Congratulations User. Your avg score is 85.66
        student_scores(90,89,78,"Raman")
        Congratulations Raman. Your avg score is 85.66
        """
    avg = (math_score+sci_score+social_score)/3
    return f"Congratulations {name} from section {section}. Your avg score is {round(avg,2)}"
```

```python
# positional argument
sc3 = student_scores2(88,76,78)
sc4 = student_scores2(88,76,45,"Raman")
sc5 = student_scores2(77,56,45,"Raman","2G")
print(sc3,sc4,sc5,sep='\n\n')
```

```python
# keyword arguments
sc6 = student_scores2(name='Raman',section='2H',math_score=90,social_score=23,sci_score=87)
print(sc6)
# Combination of positional and keyword arguments
sc7 = student_scores2(54,34,name='Raman',section='3H',social_score=88)
print(sc7)
```

**Positional arguments**: Position wise arguments that you pass while calling a function

**Keyword arguments**: You pass the data by mentioning the specific keyword

**Positional arguments are followed by keyword arguments ALWAYS (When used in combination)**

```python
def sample(username:str,emailID:str,mobile:int,payment:int)->str:
    return f"{username},{emailID},{mobile},{payment}"

# positional arguments
s1 = sample("Raman","Raman@gmail.com",5445545,9000)

# keyword arguments
s2 = sample(username="Raman",mobile=455455,payment=34000,emailID="Raman@gmail.com")
s3  =sample(username="Raman",emailID="rerr",mobile=343434,payment=23000)

# positional arguments and keyword arguments
s4  =sample("Thaman",mobile=565656,emailID="raman@gmail.com",payment=3434)

print(s1,s2,s3,s4,sep='\n\n')
```

```python
def sum_numbers(data:list)->int|float:
    '''
    Example: sum_numbers([90,'A',78.22,'B'])

    '''
    # Intialize the summation to 0
    s=0
    for i in data:
        if isinstance(i,int|float):
            # i=90 : if isinstance(90,int|float) : True
            # i= 'A' : if isinstance('A',int|float) : False
            # i = 78.22 : if isinstance(78.22,int|float) : True
            # i= 'B' : if isinstance('B',int|float) : False
            s+=i # s= 0+90 = 90
            # s= 90+78.22 = 168.22
    return s

s1 = sum_numbers([90,'A',78.22,'B'])
s2 = sum_numbers(['twst','this','sample'])
s3  = sum_numbers([89,9.87,67])
print(s1,s2,s3,sep='\n\n')
```

## Lambda Functions/Anonymous functions:

a) Anonymous Functions: Head less functions

b) On the fly, if a functionality needs to be applied, then we go for Lambda function(are not defined/stored in the memory)

syntax:

**lambda input parameter:Output parameter**

```python
# square function - generic function
def square(ip):
    return ip**2

# Lambda function/headless function
sq = lambda ip:ip**2
print(f"Lambda function,{sq(10)},Regular function {square(10)}")
```
