## Dictionary

[image.png](attachment:d4312e53-7d67-4584-a0a4-6ce4e9959a6f.png)

1) Allows to store data in key-value pairs
2) key must be unique in nature. key can be integer,string or float datatype
3) values can be duplicated in nature. values can be integers/float/strings/lists/tuples/sets
4) Dicationaries are Mutable in nature
5) Syntax: dict_name = {key1:value2,key2:value2,key3:value3}

```python
country_capital = {"Germany":"Berlin",
                   "India":"New Delhi"}
print(country_capital,type(country_capital))
```

```python
student_data = {"Raman":[56,78,90,87,65],
                "Suman":[78,98,55,42,26],
                "Baman":[56,78,65,29,85]}
print(student_data,type(student_data))
```

```python
employee_data = {"emp_id":[101,102,103,104],
                 "salary":[35000,50000,70000,90000],
                 }
print(employee_data,type(employee_data))
```

```python
student_info = {"Name":"Raman",
                "DOB":"1990-01-1",
                "Age":32,
                "Avg_scores":67.8
                }
student_info
```

```python
print(student_info)
print("=======")
print("Keys: ",student_info.keys())
print("Values : ",student_info.values())
print("Keys-Values : ",student_info.items())
```

**Access Dictionary elements:**

1) keys(): returns keys present in the dictionary
2) values() : returns values present in the dictionary
3) items() : returns key value pairs present in the dictionary

```python
student_info
```

### Accessing dictionary elements
1) dict_name[key] : will return the corresponding value. If the key is not present, it will throw an error
2) dict_name.get(key) : will return the corresponding value. If the key is not present, it will return None

```python
print(student_info["Name"],student_info["DOB"],student_info["Age"],student_info['Avg_scores'])
```

```python
student_info['Location']
```

```python
# TO fetch the values - get()
print(student_info.get("Name")) # Fetch the value corresponding to Key-Name
print(student_info.get("Location"))
print(student_info.get("Location","Not Available")) # the message can be customised
print(student_info.get("Location",{}))
```
