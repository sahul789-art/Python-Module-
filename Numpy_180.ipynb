## Statistical and Aggregation Operations
    
    prod()
    sum()
    min()
    max()
    mean()
    std()
    median() : centre most point in the data

```python
# create an array of profit of the products
profit = np.array([45000,3200,7600,12000])
print(profit)
# can you calculate the cumulative profit made by all the products
culumative_profit = np.sum(profit)
print(culumative_profit)
# can you calculate the least profit
least = np.min(profit)
print(least)
# can you calculate the maximum profit received
max_profit = np.max(profit)
print(max_profit)

# can you calculate average profit
avg_profit = np.mean(profit)
avg2 = np.average(profit)
print(avg2)
print (avg_profit)
print(round(avg_profit,2))
```

```python
# can you calculate the centre most profit area which is observed from the current dataset
print(np.median(profit))
```

```python
print(np.mean(profit))
print(np.std(profit))
print(np.var(profit))
```
