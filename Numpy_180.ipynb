**2D data: nested arrays**

```python
arr3 = np.array([["raman",87],["thaman",45],["Suman",34]])
print(arr3,type(arr3),arr3.shape,sep='\n')
```

```python
arr4 = np.array([[4,5,6],[9,10,11],[3,4,5],[1,2,3]])
print(arr4,arr4.shape,sep='\n')
# shape: (number of rows,number of columns)
# number of rows = number of sublists =
# number of columns = number of elements in each sublist
```

```python
arr5 = np.array([[2,3],[3,4,5]])
```

```python
arr_3d = np.array([[[1,2],[2,3]],[[1,2],[3,4]]])
# combination of multiple 2d sets
print(arr_3d,arr_3d.shape)
```

shape: axis0,axis1,axis
axis0: number of 2d sets

axis1: number of rows in each 2d set

axis2: number of columns in the 1d set (sublist inside 2nd set)
